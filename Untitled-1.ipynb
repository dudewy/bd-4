У какой кафедры больше всего курсов?

```python
SELECT 
    l.courseid,
    STRING_AGG(DISTINCT d.depart_name, ', ') AS departments
FROM logs l
JOIN departments d ON l.Depart = d.depart_id
GROUP BY l.courseid
HAVING COUNT(DISTINCT l.Depart) > 1;
```
